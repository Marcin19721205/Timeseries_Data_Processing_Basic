Data presparation for different distributions - to QQ plot

```python
import numpy as np                              # numerical computing + random number generation
import scipy.stats as st                        # probability distributions (normal, t, alpha-stable, skewed)

# --- Assumptions / defaults you can tweak ---
ref_col = "spx"                                 # we will match mean/std to this column (change if you want)
seed = 123                                      # reproducibility
n = len(df_comp)                                # number of rows -> new columns must have the same length

mu = df_comp[ref_col].mean()                    # target mean (taken from the reference column)
sigma = df_comp[ref_col].std(ddof=1)            # target standard deviation (sample std)

rng = np.random.default_rng(seed)               # modern RNG (better than np.random.*)

# 1) Gaussian / Normal distribution N(mu, sigma)
df_comp["Gauss"] = rng.normal(loc=mu, scale=sigma, size=n)  # normal values with matched mean/std

# 2) Student's t distribution (heavy tails), then standardized -> scaled to (mu, sigma)
df_t = 5                                        # degrees of freedom (smaller = heavier tails)
t_raw = st.t.rvs(df=df_t, size=n, random_state=seed)        # draw from t(df)
t_std = (t_raw - np.mean(t_raw)) / np.std(t_raw, ddof=1)    # standardize to mean=0, std=1
df_comp["tStudent"] = mu + sigma * t_std         # rescale to match mean/std

# 3) Alpha-stable distribution (very heavy tails), then centered -> scaled to (mu, sigma)
alpha = 1.7                                     # stability parameter (0<alpha<=2); alpha=2 -> Gaussian
beta = 0.0                                      # skewness parameter (-1..1); beta=0 -> symmetric
a_raw = st.levy_stable.rvs(alpha=alpha, beta=beta, size=n, random_state=seed)  # alpha-stable draw
a_raw = a_raw - np.median(a_raw)                # robust centering (mean may not exist for alpha<1)
a_scale = np.std(a_raw, ddof=1)                 # empirical scale (just to get similar spread)
df_comp["alfa"] = mu + sigma * (a_raw / a_scale) # rescale to match mean/std approximately

# 4) Skewed distribution (Skew-Normal), then standardized -> scaled to (mu, sigma)
skew_a = 8                                      # shape parameter; >0 right-skew, <0 left-skew
s_raw = st.skewnorm.rvs(a=skew_a, size=n, random_state=seed) # skew-normal draw
s_std = (s_raw - np.mean(s_raw)) / np.std(s_raw, ddof=1)     # standardize
df_comp["skew"] = mu + sigma * s_std            # rescale to match mean/std

# quick sanity check (optional)
print(df_comp[["Gauss", "tStudent", "alfa", "skew"]].describe())
```

```python
import numpy as np                      # arrays + numerics
import matplotlib.pyplot as plt         # plotting
import scipy.stats as st               # stats (QQ plot + PDFs)

data = df_comp["Gauss"].dropna().to_numpy()   # take Gauss column as 1D array (drop NaNs)

fig, axes = plt.subplots(1, 2, figsize=(14, 6))  # two plots side-by-side

# --- Left: QQ plot (Normality check) ---
st.probplot(data, dist="norm", plot=axes[0])     # QQ plot vs Normal distribution
axes[0].set_title("QQ Plot Gauss", fontsize=18)  # title for QQ plot
axes[0].grid(True)                                # grid for readability

# --- Right: Distribution plot (Histogram + fitted Normal PDF) ---
mu = data.mean()                                  # sample mean
sigma = data.std(ddof=1)                          # sample std (ddof=1 = sample std)
axes[1].hist(data, bins=40, density=True, alpha=0.7)  # histogram normalized to density

x = np.linspace(data.min(), data.max(), 400)      # x-grid for smooth PDF curve
axes[1].plot(x, st.norm.pdf(x, loc=mu, scale=sigma))   # fitted Normal PDF overlay

axes[1].set_title("Distribution (hist + fitted Normal)", fontsize=18)  # title
axes[1].set_xlabel("Value")                       # x-axis label
axes[1].set_ylabel("Density")                     # y-axis label
axes[1].grid(True)                                # grid

plt.tight_layout()                                # avoid overlaps
plt.show()                                        # render the figure
```

```python
import numpy as np                      # arrays + numerics
import matplotlib.pyplot as plt         # plotting
import scipy.stats as st               # stats (QQ plot + PDFs)

data = df_comp["tStudent"].dropna().to_numpy()  # take tStudent column as 1D array (drop NaNs)

fig, axes = plt.subplots(1, 2, figsize=(14, 6))  # two plots side-by-side

# --- Left: QQ plot vs Student's t distribution ---
df_t = 5                                           # degrees of freedom (must match what you used to generate the column)
st.probplot(data, dist=st.t, sparams=(df_t,), plot=axes[0])  # QQ plot against t(df_t)
axes[0].set_title("QQ Plot t-student", fontsize=18)          # title
axes[0].grid(True)                                           # grid

# --- Right: Distribution plot (Histogram + fitted t PDF) ---
loc_hat = np.mean(data)                          # location estimate (simple mean)
scale_hat = np.std(data, ddof=1)                 # scale estimate (simple std)
axes[1].hist(data, bins=40, density=True, alpha=0.7)  # histogram normalized to density

x = np.linspace(data.min(), data.max(), 500)     # x-grid for smooth PDF curve
axes[1].plot(x, st.t.pdf(x, df=df_t, loc=loc_hat, scale=scale_hat))  # t PDF overlay (approx fit)

axes[1].set_title("Distribution (hist + t PDF)", fontsize=18)  # title
axes[1].set_xlabel("Value")                       # x-axis label
axes[1].set_ylabel("Density")                     # y-axis label
axes[1].grid(True)                                # grid

plt.tight_layout()                                # avoid overlaps
plt.show()                                        # render
```

```python
import numpy as np                      # arrays + numerics
import matplotlib.pyplot as plt         # plotting
import scipy.stats as st               # stats (QQ plot + PDFs)

data = df_comp["alfa"].dropna().to_numpy()  # take alfa column as 1D array (drop NaNs)

fig, axes = plt.subplots(1, 2, figsize=(14, 6))  # two plots side-by-side

# --- Left: QQ plot vs Alpha-stable distribution ---
alpha = 1.7                                      # stability parameter (must match generation)
beta = 0.0                                       # skewness parameter (must match generation)
# NOTE: probplot expects a "distribution object"; levy_stable works, but fitting is tricky
st.probplot(data, dist=st.levy_stable, sparams=(alpha, beta), plot=axes[0])  # QQ plot vs alpha-stable
axes[0].set_title("QQ Plot alfa (alpha-stable)", fontsize=18)
axes[0].grid(True)

# --- Right: Distribution plot (Histogram + alpha-stable PDF overlay) ---
# We estimate loc/scale roughly; alpha-stable can have heavy tails and unstable moment estimates.
loc_hat = np.median(data)                          # robust location estimate
scale_hat = np.median(np.abs(data - loc_hat))      # robust scale proxy (MAD-like, not exact)

# avoid scale=0
scale_hat = max(scale_hat, 1e-12)

axes[1].hist(data, bins=60, density=True, alpha=0.7)  # histogram normalized to density

x = np.linspace(np.quantile(data, 0.01), np.quantile(data, 0.99), 600)       # avoid extreme tails for plotting
axes[1].plot(x, st.levy_stable.pdf(x, alpha, beta, loc=loc_hat, scale=scale_hat))  # PDF overlay (approx)

axes[1].set_title("Distribution (hist + alpha-stable PDF)", fontsize=18)
axes[1].set_xlabel("Value")
axes[1].set_ylabel("Density")
axes[1].grid(True)

plt.tight_layout()
plt.show()
```

```python
import numpy as np                      # arrays + numerics
import matplotlib.pyplot as plt         # plotting
import scipy.stats as st               # stats (QQ plot + PDFs)

data = df_comp["skew"].dropna().to_numpy()   # take skew column as 1D array (drop NaNs)

fig, axes = plt.subplots(1, 2, figsize=(14, 6))  # two plots side-by-side

# --- Left: QQ plot (by default vs Normal) ---
# If you want QQ vs skew-normal instead, use the alternative block below.
st.probplot(data, dist="norm", plot=axes[0])     # QQ plot against Normal distribution
axes[0].set_title("QQ Plot skew (vs Normal)", fontsize=18)
axes[0].grid(True)

# --- Right: Distribution plot (Histogram + fitted skew-normal PDF) ---
# Fit skew-normal parameters (shape a, location loc, scale) from data
a_hat, loc_hat, scale_hat = st.skewnorm.fit(data)     # maximum likelihood fit

axes[1].hist(data, bins=60, density=True, alpha=0.7)  # histogram normalized to density

x = np.linspace(np.quantile(data, 0.01), np.quantile(data, 0.99), 600)       # avoid extreme tails for plotting
axes[1].plot(x, st.skewnorm.pdf(x, a_hat, loc=loc_hat, scale=scale_hat))     # fitted skew-normal PDF overlay

axes[1].set_title("Distribution (hist + fitted skew-normal PDF)", fontsize=18)
axes[1].set_xlabel("Value")
axes[1].set_ylabel("Density")
axes[1].grid(True)

plt.tight_layout()
plt.show()

# --- OPTIONAL: QQ plot vs skew-normal instead of Normal (uncomment to use) ---
# fig, ax = plt.subplots(figsize=(7, 6))
# st.probplot(data, dist=st.skewnorm, sparams=(a_hat,), plot=ax)  # QQ vs skew-normal (approx; loc/scale not included here)
# ax.set_title("QQ Plot skew (vs skew-normal)", fontsize=18)
# ax.grid(True)
# plt.show()
```
