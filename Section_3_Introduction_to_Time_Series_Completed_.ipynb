Data presparation for different distributions - to QQ plot

```python
import numpy as np                              # numerical computing + random number generation
import scipy.stats as st                        # probability distributions (normal, t, alpha-stable, skewed)

# --- Assumptions / defaults you can tweak ---
ref_col = "spx"                                 # we will match mean/std to this column (change if you want)
seed = 123                                      # reproducibility
n = len(df_comp)                                # number of rows -> new columns must have the same length

mu = df_comp[ref_col].mean()                    # target mean (taken from the reference column)
sigma = df_comp[ref_col].std(ddof=1)            # target standard deviation (sample std)

rng = np.random.default_rng(seed)               # modern RNG (better than np.random.*)

# 1) Gaussian / Normal distribution N(mu, sigma)
df_comp["Gauss"] = rng.normal(loc=mu, scale=sigma, size=n)  # normal values with matched mean/std

# 2) Student's t distribution (heavy tails), then standardized -> scaled to (mu, sigma)
df_t = 5                                        # degrees of freedom (smaller = heavier tails)
t_raw = st.t.rvs(df=df_t, size=n, random_state=seed)        # draw from t(df)
t_std = (t_raw - np.mean(t_raw)) / np.std(t_raw, ddof=1)    # standardize to mean=0, std=1
df_comp["tStudent"] = mu + sigma * t_std         # rescale to match mean/std

# 3) Alpha-stable distribution (very heavy tails), then centered -> scaled to (mu, sigma)
alpha = 1.7                                     # stability parameter (0<alpha<=2); alpha=2 -> Gaussian
beta = 0.0                                      # skewness parameter (-1..1); beta=0 -> symmetric
a_raw = st.levy_stable.rvs(alpha=alpha, beta=beta, size=n, random_state=seed)  # alpha-stable draw
a_raw = a_raw - np.median(a_raw)                # robust centering (mean may not exist for alpha<1)
a_scale = np.std(a_raw, ddof=1)                 # empirical scale (just to get similar spread)
df_comp["alfa"] = mu + sigma * (a_raw / a_scale) # rescale to match mean/std approximately

# 4) Skewed distribution (Skew-Normal), then standardized -> scaled to (mu, sigma)
skew_a = 8                                      # shape parameter; >0 right-skew, <0 left-skew
s_raw = st.skewnorm.rvs(a=skew_a, size=n, random_state=seed) # skew-normal draw
s_std = (s_raw - np.mean(s_raw)) / np.std(s_raw, ddof=1)     # standardize
df_comp["skew"] = mu + sigma * s_std            # rescale to match mean/std

# quick sanity check (optional)
print(df_comp[["Gauss", "tStudent", "alfa", "skew"]].describe())
```

```python
import numpy as np                      # arrays + numerics
import matplotlib.pyplot as plt         # plotting
import scipy.stats as st                # stats (QQ plot + PDFs)

# ---------- Global x-axis scale + common histogram binning ----------
cols = ["Gauss", "tStudent", "alfa", "skew"]                                   # columns to compare
q_low, q_high = 0.01, 0.99                                                     # robust percentiles for common x-range

all_vals = np.concatenate([df_comp[c].dropna().to_numpy() for c in cols])      # stack all values
x_left = np.quantile(all_vals, q_low)                                          # common left x-limit
x_right = np.quantile(all_vals, q_high)                                        # common right x-limit

n_bins = 40                                                                    # same bar count everywhere
bin_edges = np.linspace(x_left, x_right, n_bins + 1)                           # fixed edges => identical bar widths

# ---------- Per-distribution settings (must match your generation) ----------
df_t = 5                                                                       # Student-t degrees of freedom
alpha = 1.7                                                                    # alpha-stable alpha
beta = 0.0                                                                     # alpha-stable beta
# For skew: we'll fit skew-normal automatically (no need to hardcode)

def plot_one(col_name: str):
    data = df_comp[col_name].dropna().to_numpy()                               # extract column as array

    fig, axes = plt.subplots(1, 2, figsize=(14, 6))                            # two plots side-by-side

    # ---------- QQ plot ----------
    if col_name == "Gauss":
        st.probplot(data, dist="norm", plot=axes[0])                           # QQ vs Normal
        axes[0].set_title("QQ Plot Gauss", fontsize=18)

    elif col_name == "tStudent":
        st.probplot(data, dist=st.t, sparams=(df_t,), plot=axes[0])            # QQ vs t(df)
        axes[0].set_title("QQ Plot t-student", fontsize=18)

    elif col_name == "alfa":
        st.probplot(data, dist=st.levy_stable, sparams=(alpha, beta), plot=axes[0])  # QQ vs alpha-stable
        axes[0].set_title("QQ Plot alfa (alpha-stable)", fontsize=18)

    elif col_name == "skew":
        st.probplot(data, dist="norm", plot=axes[0])                           # QQ vs Normal (shows skew clearly)
        axes[0].set_title("QQ Plot skew (vs Normal)", fontsize=18)

    axes[0].grid(True)                                                         # grid on QQ plot

    # ---------- Histogram (common bins) + PDF overlay ----------
    axes[1].hist(data, bins=bin_edges, density=True, alpha=0.7)                # SAME bins => SAME bar widths
    x = np.linspace(x_left, x_right, 600)                                      # COMMON x-grid

    if col_name == "Gauss":
        mu = data.mean()                                                       # mean
        sigma = data.std(ddof=1)                                               # std
        axes[1].plot(x, st.norm.pdf(x, loc=mu, scale=sigma))                   # Normal PDF

    elif col_name == "tStudent":
        loc_hat = np.mean(data)                                                # loc estimate
        scale_hat = np.std(data, ddof=1)                                       # scale estimate
        axes[1].plot(x, st.t.pdf(x, df=df_t, loc=loc_hat, scale=scale_hat))    # t PDF (approx)

    elif col_name == "alfa":
        loc_hat = np.median(data)                                              # robust loc (mean may be unstable)
        scale_hat = np.median(np.abs(data - loc_hat))                          # robust scale proxy
        scale_hat = max(scale_hat, 1e-12)                                      # avoid zero scale
        axes[1].plot(x, st.levy_stable.pdf(x, alpha, beta, loc=loc_hat, scale=scale_hat))  # alpha-stable PDF

    elif col_name == "skew":
        a_hat, loc_hat, scale_hat = st.skewnorm.fit(data)                      # fit skew-normal parameters
        axes[1].plot(x, st.skewnorm.pdf(x, a_hat, loc=loc_hat, scale=scale_hat))  # skew-normal PDF

    axes[1].set_title(f"Distribution: {col_name}", fontsize=18)                # title
    axes[1].set_xlabel("Value")                                                # x label
    axes[1].set_ylabel("Density")                                              # y label
    axes[1].grid(True)                                                         # grid
    axes[1].set_xlim(x_left, x_right)                                          # enforce COMMON x-limits

    plt.tight_layout()
    plt.show()

# ---------- Run all four ----------
for c in cols:
    plot_one(c)
```

```python
import numpy as np                      # arrays + numerics
import matplotlib.pyplot as plt         # plotting
import scipy.stats as st                # stats (QQ plot + PDFs)

# ---------- Global x-axis scale + common histogram binning ----------
cols = ["Gauss", "tStudent", "alfa", "skew"]                                   # columns to compare
q_low, q_high = 0.01, 0.99                                                     # robust percentiles for common x-range

all_vals = np.concatenate([df_comp[c].dropna().to_numpy() for c in cols])      # stack all values
x_left = np.quantile(all_vals, q_low)                                          # common left x-limit
x_right = np.quantile(all_vals, q_high)                                        # common right x-limit

n_bins = 40                                                                    # same bar count everywhere
bin_edges = np.linspace(x_left, x_right, n_bins + 1)                           # fixed edges => identical bar widths

# ---------- Per-distribution settings (must match your generation) ----------
df_t = 5                                                                       # Student-t degrees of freedom
alpha = 1.7                                                                    # alpha-stable alpha
beta = 0.0                                                                     # alpha-stable beta
# For skew: we'll fit skew-normal automatically (no need to hardcode)

def plot_one(col_name: str):
    data = df_comp[col_name].dropna().to_numpy()                               # extract column as array

    fig, axes = plt.subplots(1, 2, figsize=(14, 6))                            # two plots side-by-side

    # ---------- QQ plot ----------
    if col_name == "Gauss":
        st.probplot(data, dist="norm", plot=axes[0])                           # QQ vs Normal
        axes[0].set_title("QQ Plot Gauss", fontsize=18)

    elif col_name == "tStudent":
        st.probplot(data, dist=st.t, sparams=(df_t,), plot=axes[0])            # QQ vs t(df)
        axes[0].set_title("QQ Plot t-student", fontsize=18)

    elif col_name == "alfa":
        st.probplot(data, dist=st.levy_stable, sparams=(alpha, beta), plot=axes[0])  # QQ vs alpha-stable
        axes[0].set_title("QQ Plot alfa (alpha-stable)", fontsize=18)

    elif col_name == "skew":
        st.probplot(data, dist="norm", plot=axes[0])                           # QQ vs Normal (shows skew clearly)
        axes[0].set_title("QQ Plot skew (vs Normal)", fontsize=18)

    axes[0].grid(True)                                                         # grid on QQ plot

    # ---------- Histogram (common bins) + PDF overlay ----------
    axes[1].hist(data, bins=bin_edges, density=True, alpha=0.7)                # SAME bins => SAME bar widths
    x = np.linspace(x_left, x_right, 600)                                      # COMMON x-grid

    if col_name == "Gauss":
        mu = data.mean()                                                       # mean
        sigma = data.std(ddof=1)                                               # std
        axes[1].plot(x, st.norm.pdf(x, loc=mu, scale=sigma))                   # Normal PDF

    elif col_name == "tStudent":
        loc_hat = np.mean(data)                                                # loc estimate
        scale_hat = np.std(data, ddof=1)                                       # scale estimate
        axes[1].plot(x, st.t.pdf(x, df=df_t, loc=loc_hat, scale=scale_hat))    # t PDF (approx)

    elif col_name == "alfa":
        loc_hat = np.median(data)                                              # robust loc (mean may be unstable)
        scale_hat = np.median(np.abs(data - loc_hat))                          # robust scale proxy
        scale_hat = max(scale_hat, 1e-12)                                      # avoid zero scale
        axes[1].plot(x, st.levy_stable.pdf(x, alpha, beta, loc=loc_hat, scale=scale_hat))  # alpha-stable PDF

    elif col_name == "skew":
        a_hat, loc_hat, scale_hat = st.skewnorm.fit(data)                      # fit skew-normal parameters
        axes[1].plot(x, st.skewnorm.pdf(x, a_hat, loc=loc_hat, scale=scale_hat))  # skew-normal PDF

    axes[1].set_title(f"Distribution: {col_name}", fontsize=18)                # title
    axes[1].set_xlabel("Value")                                                # x label
    axes[1].set_ylabel("Density")                                              # y label
    axes[1].grid(True)                                                         # grid
    axes[1].set_xlim(x_left, x_right)                                          # enforce COMMON x-limits

    plt.tight_layout()
    plt.show()

# ---------- Run all four ----------
for c in cols:
    plot_one(c)
```
