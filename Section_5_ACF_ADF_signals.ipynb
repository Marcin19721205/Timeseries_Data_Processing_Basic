#ADF

```python
import statsmodels.tsa.stattools as sts  # ADF test lives here

# --- columns to test (use exactly the columns present in df) ---
cols = list(df.columns)                 # all signals / columns in df
alpha = 0.05                            # significance level for decisions

print(f"Running ADF for {len(cols)} columns: {cols}")

for col in cols:
    print("\n" + "=" * 80)
    print(f"ADF test | column: {col}")

    # --- run ADF test (drop NaNs because ADF hates missing values) ---
    series = df[col].dropna()                                           # clean series (no NaNs)
    adf_out = sts.adfuller(series, autolag="AIC")                        # ADF + automatic lag selection by AIC

    # --- unpack results ---
    adf_stat = adf_out[0]                                               # test statistic
    p_value  = adf_out[1]                                               # p-value
    used_lags = adf_out[2]                                              # number of lags used
    n_obs = adf_out[3]                                                  # number of observations used
    crit = adf_out[4]                                                   # critical values dict: 1%, 5%, 10%
    icbest = adf_out[5]                                                 # best information criterion value (from autolag)

    # --- print raw results ---
    print(f"ADF statistic: {adf_stat:.6f}")                             # more negative -> stronger evidence vs unit root
    print(f"p-value:       {p_value:.6g}")                              # key decision number
    print(f"used lags:     {used_lags}")                                # how many lagged differences included
    print(f"n obs:         {n_obs}")                                    # effective sample size
    print(f"critical values: 1%={crit['1%']:.6f}, 5%={crit['5%']:.6f}, 10%={crit['10%']:.6f}")
    print(f"IC best:       {icbest:.6f}")                               # chosen by autolag criterion

    # --- interpret stationarity ---
    if p_value < alpha:
        print(f"Decision @ {alpha}: REJECT H0 (unit root) -> series is likely STATIONARY.")
    else:
        print(f"Decision @ {alpha}: FAIL TO REJECT H0 -> series is likely NON-STATIONARY (unit root).")

    # --- optional: interpretation using critical values (extra sanity check) ---
    if adf_stat < crit["5%"]:
        print("Sanity check: ADF statistic is below the 5% critical value -> supports stationarity.")
    else:
        print("Sanity check: ADF statistic is NOT below the 5% critical value -> supports non-stationarity.")
```

ADF with options

```python
import statsmodels.tsa.stattools as sts  # ADF test (adfuller)

# --- columns to test (use exactly the columns present in df) ---
cols = list(df.columns)                 # all signals / columns in df
alpha = 0.05                            # significance level

# Meanings + what "stationary" would mean under each option
reg_info = {
    "c":  {
        "meaning": "constant only (drift) included",
        "stationary_means": "stationary around a constant mean (mean-reverting to a level)"
    },
    "ct": {
        "meaning": "constant + linear time trend included",
        "stationary_means": "stationary around a deterministic linear trend (reverting to a trend line)"
    },
    "n":  {
        "meaning": "no constant, no trend included",
        "stationary_means": "stationary around zero (no drift, no trend) — strong assumption"
    }
}

print(f"Running ADF for {len(cols)} columns (regressions: c, ct, n):\n{cols}")

for col in cols:
    series = df[col].dropna()  # clean series (no NaNs)

    print("\n" + "#" * 90)
    print(f"COLUMN: {col} | n={len(series)}")
    print("#" * 90)

    for reg in ["c", "ct", "n"]:
        stat, pval, used_lags, nobs, crit, icbest = sts.adfuller(series, regression=reg, autolag="AIC")

        print("\n" + "=" * 70)
        print(f"ADF option regression='{reg}'")
        print(f"  Meaning: {reg_info[reg]['meaning']}")
        print(f"  If stationary (H1), it would mean: {reg_info[reg]['stationary_means']}")
        print("-" * 70)

        print(f"  ADF statistic : {stat:.6f}")
        print(f"  p-value       : {pval:.6g}")
        print(f"  used lags     : {used_lags}")
        print(f"  n obs         : {nobs}")
        print(f"  critical vals : 1%={crit['1%']:.6f}, 5%={crit['5%']:.6f}, 10%={crit['10%']:.6f}")
        print(f"  icbest        : {icbest:.6f}")

        # Decision logic for ADF:
        # H0: unit root (non-stationary)
        # H1: stationary (as described above, depending on reg)
        if pval < alpha:
            print(f"  Decision @ {alpha}: REJECT H0 (unit root) -> likely STATIONARY.")
            print(f"  Interpretation: {reg_info[reg]['stationary_means']}")
        else:
            print(f"  Decision @ {alpha}: FAIL TO REJECT H0 -> likely NON-STATIONARY (unit root).")
            print("  Interpretation: test did not provide evidence for stationarity under this specification.")
```

```python
import numpy as np
import pandas as pd
from statsmodels.tsa.stattools import acf
from scipy.signal import find_peaks

cols = list(df.columns)
max_lags = 800
alpha = 0.05
min_lag = 50                 # <-- ignore tiny lags like 1..49 (tune: 50/100)

rows = []

for col in cols:
    x = df[col].dropna().to_numpy()

    acf_vals, confint = acf(x, nlags=max_lags, alpha=alpha, fft=True)

    lags = np.arange(1, max_lags + 1)
    a = acf_vals[1:]

    # CI centered around 0
    ci_low = confint[1:, 0] - acf_vals[1:]
    ci_high = confint[1:, 1] - acf_vals[1:]

    # significance mask
    sig = (a < ci_low) | (a > ci_high)

    # search region
    idx0 = min_lag - 1
    a2 = a[idx0:]
    l2 = lags[idx0:]
    sig2 = sig[idx0:]

    # find peaks in |ACF| (works for both positive and negative peaks)
    peaks, _ = find_peaks(np.abs(a2))

    # keep only significant peaks
    peaks = [p for p in peaks if sig2[p]]

    if len(peaks) > 0:
        # pick the strongest peak by |ACF|
        best_p = max(peaks, key=lambda p: abs(a2[p]))
        dom_lag = int(l2[best_p])
        dom_acf = float(a2[best_p])
    else:
        dom_lag = None
        dom_acf = None

    rows.append({
        "column": col,
        "dominant_lag_from_min_lag": dom_lag,
        "acf_at_dominant_lag": dom_acf,
        "min_lag_used": min_lag
    })

summary2 = pd.DataFrame(rows).sort_values(["dominant_lag_from_min_lag","column"])
print(summary2.to_string(index=False))
```

ACF - Autocorrelation

```python
import numpy as np
import pandas as pd
import plotly.graph_objects as go
from statsmodels.tsa.stattools import acf
from scipy.signal import find_peaks

# =========================
# SETTINGS
# =========================
cols = list(df.columns)
nlags = 800                 # increase/decrease depending on expected period
alpha = 0.05                # 95% CI
min_lag_for_period = 50     # ignore tiny lags when searching dominant period
mode = "raw"               # "raw" or "diff" (diff is usually "proper" for non-stationary levels)

# =========================
# PREPROCESS
# =========================
def preprocess_series(s: pd.Series, mode: str) -> np.ndarray:
    s = s.dropna().copy()

    if mode == "raw":
        x = s.to_numpy()
    elif mode == "diff":
        x = s.diff().dropna().to_numpy()             # detrending by 1st difference
    else:
        raise ValueError("mode must be 'raw' or 'diff'")

    # standardize (doesn't change ACF shape, just keeps numerics tidy)
    x = (x - np.mean(x)) / (np.std(x, ddof=1) + 1e-12)
    return x

# =========================
# DOMINANT LAG PICKER
# =========================
def dominant_lag(acf_vals, confint, min_lag: int):
    a = acf_vals[1:]                                 # lags 1..N
    lags = np.arange(1, len(a) + 1)

    ci_low = confint[1:, 0] - acf_vals[1:]          # CI around 0
    ci_high = confint[1:, 1] - acf_vals[1:]
    sig = (a < ci_low) | (a > ci_high)              # outside CI = significant

    idx0 = max(min_lag - 1, 0)
    a2 = a[idx0:]
    l2 = lags[idx0:]
    sig2 = sig[idx0:]

    # peaks in |ACF|, keep only significant ones
    peaks, _ = find_peaks(np.abs(a2))
    peaks = [p for p in peaks if sig2[p]]

    if len(peaks) == 0:
        return None, None

    best_p = max(peaks, key=lambda p: abs(a2[p]))
    return int(l2[best_p]), float(a2[best_p])

# =========================
# RUN ALL COLUMNS: PLOTS + SUMMARY
# =========================
summary = []

for col in cols:
    x = preprocess_series(df[col], mode=mode)

    acf_vals, confint = acf(x, nlags=nlags, alpha=alpha, fft=True)

    dom_lag, dom_val = dominant_lag(acf_vals, confint, min_lag=min_lag_for_period)
    summary.append({
        "column": col,
        "mode": mode,
        "dominant_lag": dom_lag,
        "acf_at_dominant_lag": dom_val,
    })

    # --- Build Plotly ACF chart ---
    lags = np.arange(1, nlags + 1)
    a = acf_vals[1:]
    ci_low = confint[1:, 0] - acf_vals[1:]
    ci_high = confint[1:, 1] - acf_vals[1:]

    # stems
    x_stem = np.repeat(lags, 3)
    y_stem = np.ravel(np.column_stack([np.zeros_like(a), a, np.full_like(a, np.nan)]))

    fig = go.Figure()

    # CI band
    fig.add_trace(go.Scatter(
        x=np.concatenate([lags, lags[::-1]]),
        y=np.concatenate([ci_high, ci_low[::-1]]),
        fill="toself",
        name="95% CI",
        line=dict(width=0),
        hoverinfo="skip",
        opacity=0.25
    ))

    # zero line
    fig.add_trace(go.Scatter(
        x=[0, nlags], y=[0, 0],
        mode="lines",
        name="0 line",
        hoverinfo="skip",
        line=dict(dash="dash")
    ))

    # stems + points
    fig.add_trace(go.Scatter(x=x_stem, y=y_stem, mode="lines", name="ACF", line=dict(width=2)))
    fig.add_trace(go.Scatter(x=lags, y=a, mode="markers", name="ACF points", marker=dict(size=5)))

    # mark dominant lag (if found)
    if dom_lag is not None:
        fig.add_vline(x=dom_lag, line_dash="dot", line_width=2)
        title_extra = f" | dominant_lag={dom_lag} (ACF={dom_val:.3f})"
    else:
        title_extra = " | dominant_lag=NONE"

    fig.update_layout(
        template="plotly_dark",
        title=f"ACF | {col} | mode={mode} | nlags={nlags}{title_extra}",
        height=550,
        hovermode="x unified",
        showlegend=False
    )
    fig.update_xaxes(title_text="Lag (k)")
    fig.update_yaxes(title_text="Autocorrelation", range=[-1, 1])

    fig.show()

# --- summary table ---
summary_df = pd.DataFrame(summary).sort_values(["dominant_lag", "column"])
print("\n" + "="*90)
print("SUMMARY (dominant lag estimated from ACF, ignoring small lags)")
print("="*90)
print(summary_df.to_string(index=False))
```

Multiplicative model

```python
from statsmodels.tsa.seasonal import seasonal_decompose  # decomposition
from plotly.subplots import make_subplots                # subplot layout
import plotly.graph_objects as go                        # Plotly traces
import numpy as np                                       # numeric utils (for safety checks)

# --- settings ---
cols = list(df.columns)  # all columns in df

# Define period PER column (edit these values as you want)
period_map = {
    "s2_0s_1_NoNoise":       625,
    "s2_0s_1_001Noise":      625,
    "s2_0s_1_0001Noise":     625,
    "s2_0s_1_00001Noise":    625,
    "s2_01s_1_NoNoise":      625,
    "s2_01s_1_001Noise":     625,
    "s2_01s_1_0001Noise":    625,
    "s2_01s_1_00001Noise":   625,
}

default_period = 625  # used if a column is missing from period_map

for col in cols:
    period = period_map.get(col, default_period)  # pick period for this column

    print("\n" + "=" * 90)
    print(f"MULTIPLICATIVE MODEL | Seasonal Decomposition | column: {col} | period={period}")
    print("=" * 90)

    # --- prepare a clean, regular time series ---
    series = df[col].copy()                               # take the target series
    if series.index.inferred_freq is None:
        series = series.asfreq("D")                       # your data is daily; use "D" unless you KNOW it's business days

    series = series.interpolate("time").ffill().bfill()   # fill gaps (decompose needs no NaNs)

    # --- multiplicative model requires strictly positive values ---
    # If you have zeros/negatives, shift the series upward (otherwise decomposition can fail or be meaningless).
    min_val = float(series.min())
    shift = 0.0
    if min_val <= 0:
        shift = abs(min_val) + 1e-6                       # small epsilon to keep it strictly > 0
        series = series + shift

    # --- decomposition ---
    dec = seasonal_decompose(series, model="multiplicative", period=period)

    # --- Plotly figure: 4 panels stacked ---
    fig = make_subplots(
        rows=4, cols=1, shared_xaxes=True, vertical_spacing=0.03,
        subplot_titles=("Observed", "Trend", "Seasonal", "Residual-model error")
    )

    fig.add_trace(go.Scatter(x=series.index, y=dec.observed, name="Observed", mode="lines"), row=1, col=1)
    fig.add_trace(go.Scatter(x=series.index, y=dec.trend,    name="Trend",    mode="lines"), row=2, col=1)
    fig.add_trace(go.Scatter(x=series.index, y=dec.seasonal, name="Seasonal", mode="lines"), row=3, col=1)
    fig.add_trace(go.Scatter(x=series.index, y=dec.resid,    name="Residual", mode="lines"), row=4, col=1)

    title_extra = f" | shift={shift:.6g}" if shift > 0 else ""  # show if we had to shift for positivity

    fig.update_layout(
        title=f"Seasonal Decomposition (multiplicative) | column={col} | period={period}{title_extra}",
        height=900,
        showlegend=False,
        hovermode="x unified",
        template="plotly_dark"
    )

    fig.update_xaxes(rangeslider=dict(visible=True), row=4, col=1)  # range slider at the bottom
    fig.show()
```
